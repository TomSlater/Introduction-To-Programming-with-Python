### 1.2 For Loops

A **for loop** lets us repeat a task many times. The most simple task is just to print something multiple times, for example we could print a set of numbers:

```python
for n in range(1, 6):
    print(n)
```

In the code above, the `for` loop iterates over an "iterable" generated by the `range` function. This is similar to a list that contains the numbers 1 to 5. In each loop, the value of `n` changes. Here, we use `n` but you can use any name instead!

```python
for number in range(1, 6):
    print(number)
```

Many Python objects can be iterated over, including lists. We could rewrite the code above explicitly using a list instead as below.

```python
example_list = [1, 2, 'three', 4.0, 5]

for n in example_list:
    print(n)
```

### 1.3 Functions

A **function** allows you to repeat code on different input variables (called arguments). We've already seen some different functions like `print`!

In chemistry, we often want to reuse small pieces of code — for example, calculating the molar mass of molecules containing hydrogen and oxygen:

```python
def molar_mass(h_count, o_count):
    h_mass = 1.008
    o_mass = 15.999
    return(h_count * h_mass + o_count * o_mass)

print("Molar mass of H₂O:", molar_mass(2,1), "g/mol")
```

The `def` statement tells the interpreter that a function is being defined. The function name follows `def`, with the arguments of the function in brackets following the function name. The first line of the function definition must finish with a colon (:).

The function definition follows the colon. All code within the function must be indented for the interpreter to understand that it is part of the function. You can use any indentation, but the most commonly used convention is that indentation should be four spaces.

In the example above, the function has a return statement. Everything included in the brackets next to `return` is the output of the function. A function doesn't have to output anything; if you don't include a return statement the function will return `None`. We could modify the function above to just print the molar mass without returning the value, i.e.:

```python
def molar_mass(h_count, o_count):
    h_mass = 1.008
    o_mass = 15.999
    print("Molar mass:", h_count * h_mass + o_count * o_mass, "g/mol")

molar_mass(2,1)
```

### 1.4 Working beyond basic Python

We've already seen some basic maths operators in the code above, for example we could execute the following:

```python
print(2+5)
print(9-3)
print(10*3)
print(12/4)
print(10**3)
```

However, we soon run out of functions in standard Python. If we wanted to take the lograithm of a number, we would potentially have to code this ourselves. Thankfully, we don't have to do this as someone else has already done this and added it to a library that we can import.

To import the `math` library (aka module) we use the `import` command.

```python
import math
```

There are a huge number of different libraries out there and not all of them are immediately available to us, but all of the ones you need should be available to import.

The math library includes a number of functions, including one specifically for calculating logarithms. Functions that belong to an object, in this case the imported math object, are speicifically called methods. To access a method of an object, we use `.`. For example:

```python
math.log(2)
```

The `.` tells the interpreter to look inside `math` for a method called `log`. We'll use a lot of methods over this course, so do remember this!

It's also possible to directly import a method from a library, so that none of the other methods are available to use. For example:

```python
from math import log
```

It's also possible to import a library or method, but give it a different name for your use. For example:

```python
from math import log as loggy
```

```python
loggy(2)
```

## 2. Commenting and Style
<a id='section2'></a>

When programming in any language, it's really important to make your code as easy to understand as possible.

One straightforward way to do this is to leave comments in your code. A comment is a section of a code cell that is not run as code and can be indicated either by a `#` (which will comment out the rest of a line) or by a pair of `"""` (which will comment out everything in between the pair). The official correct style is to use `#`. For example:

```python
print('Hello') #Code to print Hello.

"""The above code will print Hello."""
```

Above, we've mentioned a correct style of coding in Python. There is no one official correct style, but many projects follow the [PEP8 style guide](https://peps.python.org/pep-0008/).

The PEP8 style guide is quite long, but to summarise the main points of use for us:
- Indentations should always be four spaces.
- Comments should use `#` and generally be full English sentences.
- Variable and function names should be all lower case with words separated by underscores.
- There are a lot of rules on the use of spaces and blank lines. We won't enforce these, but do read them if you are interested.

Your assessed tasks will have marks for following the PEP8 rules, so please do make sure you know the basics above.
